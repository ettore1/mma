## Passo a passo (exemplo: Exercício 1)
Cada linha é um *passo* = resolução da relaxação de um nó, na ordem da busca em profundidade.

```python
res = bb.branch_and_bound(problems_p1.ex1(), max_steps=200)
for n in sorted([x for x in res['nodes'] if x.step], key=lambda a: a.step):
    z = f'{n.z:.3f}' if n.z is not None else '--'
    print(f"passo {n.step:>2} | N{n.id:<2} | ramo: {n.branch_txt:<10} | z_LP={z:<8} | {n.status:<11} | {n.note}")
```
